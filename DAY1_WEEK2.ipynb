```python
from FlagEmbedding import FlagAutoModel

model = FlagAutoModel.from_finetuned('BAAI/bge-m3',
                                      use_fp16=True)
```

```python
sentences = [
    "Я купил новый современный автомобиль.",
    "Вчера у меня сломалась старая машина.",

    "Робин Гуд натянул тетиву и выстрелил из лука.",
    "Для приготовления супа нужно мелко нарезать репчатый лук.",

    "Стороны подписали двусторонний договор купли-продажи.",
    "Завтра синоптики обещают дождливую и ветреную погоду."
]
```

```python
sentences2 = [
    "автомобиль",
    "машина",

    "лук репчатый",
    "лук боевой",

    "договор",
    "погода"
]
```

```python
another_output = model.encode(sentences2)
another_embeddings = another_output['dense_vecs']
```

```python
output = model.encode(sentences)
embeddings = output['dense_vecs']
```

```python
def cosine_similarity(a, b):
    return sum([a_i * b_i for a_i, b_i in zip(a, b)]) / (
        sum([a_i ** 2 for a_i in a]) ** 0.5 * sum([b_i ** 2 for b_i in b]) ** 0.5)
```

```python
from sentence_transformers import util

sim_synonyms1 = util.cos_sim(embeddings[0], embeddings[1])
sim_homonyms1 = util.cos_sim(embeddings[2], embeddings[3])
sim_unrelated1 = util.cos_sim(embeddings[4], embeddings[5])

sim_synonyms = cosine_similarity(embeddings[0], embeddings[1])
sim_homonyms = cosine_similarity(embeddings[2], embeddings[3])
sim_unrelated = cosine_similarity(embeddings[4], embeddings[5])

print(f"Синонимы (вручную): {sim_synonyms:.4f}")
print(f"Синонимы (с библиотеки): {sim_synonyms1.item():.4f}")

print(f"Омонимы (вручную): {sim_homonyms:.4f} ")
print(f"Омонимы (с библиотеки):  {sim_homonyms1.item():.4f}")

print(f"Несвязанные (вручную): {sim_unrelated:.4f} ")
print(f"Несвязанные (с библиотеки): {sim_unrelated1.item():.4f}")
```
