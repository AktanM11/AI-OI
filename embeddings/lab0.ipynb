```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer('google/embeddinggemma-300m')
```

```python
import numpy as np

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

```python
query = "он сделал доброе дело"
query_embedding = model.encode_query(query)
key_1 = "справедливость восторжествовала"
key_1_embedding = model.encode_document(key_1)
cosine_similarity(query_embedding, key_1_embedding)
```

```python
key_2 = "зло не спит"
key_2_embedding = model.encode_document(key_2)
cosine_similarity(query_embedding, key_2_embedding)
```

```python
model.similarity(query_embedding, key_1_embedding)
```

```python
key_3 = "вчера мы мыли ковер"
key_3_embedding = model.encode_document(key_3)
cosine_similarity(query_embedding, key_3_embedding)
```

```python
omonim_1 = "Лучник натянул лук перед выстрелом."
omonim_2 = "Лук добавили в овощной салат."
omonim_1_embedding = model.encode_query(omonim_1)
omonim_2_embedding = model.encode_document(omonim_2)
cosine_similarity(omonim_1_embedding, omonim_2_embedding)
```

```python
omonim_1_embedding = model.encode_query(omonim_1)
omonim_2_embedding = model.encode_query(omonim_2)
cosine_similarity(omonim_1_embedding, omonim_2_embedding)
```

```python
test_query = "ты модель которая ищет овощи"
test_query_embedding = model.encode_query(test_query)
model.similarity(test_query_embedding, omonim_1_embedding), model.similarity(test_query_embedding, omonim_2_embedding)
```

```python
test2_query = "ты модель которая ищет стрелков"
test2_query_embedding = model.encode_query(test2_query)
model.similarity(test2_query_embedding, omonim_1_embedding), model.similarity(test2_query_embedding, omonim_2_embedding)
```
